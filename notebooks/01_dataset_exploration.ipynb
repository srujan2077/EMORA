```python
from sklearn.model_selection import train_test_split

video_ids = sorted(video_ids)

train_ids, temp_ids = train_test_split(
    video_ids,
    test_size=0.30,
    random_state=42
)

val_ids, test_ids = train_test_split(
    temp_ids,
    test_size=0.50,
    random_state=42
)

print("Train videos:", len(train_ids))
print("Validation videos:", len(val_ids))
print("Test videos:", len(test_ids))
```

```python
train_ids = set(train_ids)
val_ids = set(val_ids)
test_ids = set(test_ids)

train_segments = []
val_segments = []
test_segments = []

for segment_id in labels.data.keys():
    video_id = segment_id.rsplit("[", 1)[0]

    if video_id in train_ids:
        train_segments.append(segment_id)
    elif video_id in val_ids:
        val_segments.append(segment_id)
    elif video_id in test_ids:
        test_segments.append(segment_id)

print("Train segments:", len(train_segments))
print("Validation segments:", len(val_segments))
print("Test segments:", len(test_segments))

print("\nOverlap checks:")
print("Train ∩ Validation:", len(set(train_ids) & set(val_ids)))
print("Train ∩ Test:", len(set(train_ids) & set(test_ids)))
print("Validation ∩ Test:", len(set(val_ids) & set(test_ids)))
```

```python
import torch
import torch.nn as nn


class TextEncoder(nn.Module):
    def __init__(self, input_dim=300, hidden_dim=128, output_dim=128):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            batch_first=True,
            bidirectional=True
        )

        self.projection = nn.Linear(
            hidden_dim * 2,
            output_dim
        )

    def forward(self, x):
        _, (hidden, _) = self.lstm(x)

        forward_hidden = hidden[-2]
        backward_hidden = hidden[-1]

        combined = torch.cat(
            [forward_hidden, backward_hidden],
            dim=1
        )

        return self.projection(combined)
```

```python
import torch
import torch.nn as nn


class TextEncoder(nn.Module):
    def __init__(self, input_dim=300, hidden_dim=128, output_dim=128):
        super().__init__()

        self.lstm = nn.LSTM(
            input_size=input_dim,
            hidden_size=hidden_dim,
            batch_first=True,
            bidirectional=True
        )

        self.projection = nn.Linear(
            hidden_dim * 2,
            output_dim
        )

    def forward(self, x):
        _, (hidden, _) = self.lstm(x)

        forward_hidden = hidden[-2]
        backward_hidden = hidden[-1]

        combined = torch.cat(
            [forward_hidden, backward_hidden],
            dim=1
        )

        return self.projection(combined)


text_encoder = TextEncoder()

print(text_encoder)
```
